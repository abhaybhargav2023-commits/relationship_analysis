```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Student_ID": list(range(101, 131)),
    "Study_Hours": [
        1.0, 1.5, 2.0, 2.2, 2.5,
        3.0, 3.2, 3.5, 4.0, 4.2,
        4.5, 5.0, 5.2, 5.5, 6.0,
        6.2, 6.5, 7.0, 7.2, 7.5,
        8.0, 8.2, 8.5, 9.0, 9.2,
        9.5, 10.0, 10.2, 11.0, 12.0
    ],
    "Exam_Score": [
        38, 42, 45, 43, 49,
        48, 54, 52, 58, 55,
        61, 60, 64, 63, 68,
        66, 72, 70, 74, 73,
        79, 77, 83, 81, 85,
        84, 89, 87, 93, 96
    ]
}

df = pd.DataFrame(data)

df.to_csv("student_performance.csv", index=False)

print("Dataset preview:")
display(df.head())

print("Dataset shape:", df.shape)
```

```python
print("Dataset information:")
df.info()

print("\nDescriptive statistics:")
display(df[["Study_Hours", "Exam_Score"]].describe())

print("\nMissing values:")
display(df[["Study_Hours", "Exam_Score"]].isnull().sum())
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df["Study_Hours"],
    df["Exam_Score"],
    color="#2563EB",
    edgecolor="white",
    linewidth=0.8,
    s=90,
    alpha=0.85
)

plt.title(
    "Relationship Between Study Hours and Exam Scores",
    fontsize=15,
    fontweight="bold"
)
plt.xlabel("Study Hours", fontsize=12)
plt.ylabel("Exam Score", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.25)
plt.tight_layout()

plt.savefig(
    "study_hours_scatter.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
correlation = df["Study_Hours"].corr(df["Exam_Score"])

print(f"Pearson correlation coefficient: {correlation:.3f}")

if correlation >= 0.7:
    print("Interpretation: Strong positive linear relationship.")
elif correlation >= 0.3:
    print("Interpretation: Moderate positive linear relationship.")
elif correlation > -0.3:
    print("Interpretation: Weak linear relationship.")
elif correlation > -0.7:
    print("Interpretation: Moderate negative linear relationship.")
else:
    print("Interpretation: Strong negative linear relationship.")
```

```python
x = df["Study_Hours"]
y = df["Exam_Score"]

slope, intercept = __import__("numpy").polyfit(x, y, 1)
trend_line = slope * x + intercept

plt.figure(figsize=(10, 6))

plt.scatter(
    x, y,
    color="#2563EB",
    edgecolor="white",
    s=90,
    alpha=0.85,
    label="Student observations"
)

plt.plot(
    x, trend_line,
    color="#F97316",
    linewidth=2.5,
    label="Linear trend line"
)

plt.title(
    "Study Hours vs Exam Scores",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.legend(frameon=False)
plt.grid(True, linestyle="--", alpha=0.25)
plt.tight_layout()

plt.show()
```

```python
import numpy as np

x = df["Study_Hours"]
y = df["Exam_Score"]

correlation = x.corr(y)
slope, intercept = np.polyfit(x, y, 1)
trend_line = slope * x + intercept

fig, ax = plt.subplots(figsize=(12, 7))
fig.patch.set_facecolor("#F1F5F9")
ax.set_facecolor("white")

# Scatter plot
ax.scatter(
    x, y,
    s=110,
    color="#2563EB",
    edgecolor="white",
    linewidth=1.2,
    alpha=0.9,
    label="Student observations",
    zorder=3
)

# Trend line
ax.plot(
    x, trend_line,
    color="#F97316",
    linewidth=2.8,
    label="Linear trend line",
    zorder=2
)

ax.set_title(
    "STUDY HOURS vs EXAM SCORES",
    fontsize=19,
    fontweight="bold",
    color="#172554",
    pad=20
)

ax.set_xlabel("Study Hours", fontsize=12, labelpad=10)
ax.set_ylabel("Exam Score", fontsize=12, labelpad=10)

ax.grid(True, linestyle="--", alpha=0.22)
ax.legend(frameon=False, fontsize=10, loc="upper left")

for spine in ["top", "right"]:
    ax.spines[spine].set_visible(False)

fig.text(
    0.13, 0.02,
    f"Students: {len(df)}   |   Pearson Correlation: {correlation:.3f}"
    f"   |   Mean Score: {y.mean():.1f}",
    fontsize=11,
    color="#475569"
)

fig.text(
    0.13, -0.015,
    "Sample data for educational purposes • Correlation does not imply causation",
    fontsize=9,
    color="#64748B"
)

plt.tight_layout(rect=[0, 0.06, 1, 1])

plt.savefig(
    "relationship_analysis_dashboard.png",
    dpi=300,
    bbox_inches="tight",
    facecolor=fig.get_facecolor()
)

plt.show()
```
